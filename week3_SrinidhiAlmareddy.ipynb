Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
```

Load the Dataset

```python
df = pd.read_csv("Country-data.csv")
df.head()
```

Basic Information

```python
print("Shape:", df.shape)

df.info()
```

Strip Whitespace from Column Names

```python
df.columns = df.columns.str.strip()

print(df.columns.tolist())
```

Remove Duplicate Records

```python
print("Duplicates Before:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Duplicates After:", df.duplicated().sum())
```

Force Numeric Types and Handle Missing Values

```python
numeric_cols = df.columns.drop('country')

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
    df[col].fillna(df[col].median(), inplace=True)

df.isnull().sum()
```

Remove Country Identifier

```python
X = df.drop(columns=['country'])

X.head()
```

Scale Features

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print(X_scaled.shape)
```

Elbow Method (K = 2 to 10)

```python
inertia = []

K = range(2, 11)

for k in K:
    kmeans = KMeans(n_clusters=k,
                    random_state=42,
                    n_init=10)

    kmeans.fit(X_scaled)

    inertia.append(kmeans.inertia_)
```

Plot Elbow Curve

```python
plt.figure(figsize=(8,5))

plt.plot(K, inertia, marker='o')

plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.title("Elbow Method")

plt.grid(True)

plt.show()
```

Train K-Means (best_k = 3)

```python
best_k = 3

kmeans = KMeans(n_clusters=best_k,
                random_state=42,
                n_init=10)

clusters = kmeans.fit_predict(X_scaled)

df["KMeans_Cluster"] = clusters

df[['country','KMeans_Cluster']].head()
```

Silhouette Score

```python
score = silhouette_score(X_scaled, clusters)

print("Silhouette Score:", round(score, 4))
```

DBSCAN Clustering

```python
dbscan = DBSCAN(eps=1.5,
                min_samples=5)

db_labels = dbscan.fit_predict(X_scaled)

df["DBSCAN_Cluster"] = db_labels

print("DBSCAN Clusters:",
      np.unique(db_labels))
```

PCA for 2D Projection

```python
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

pca_df = pd.DataFrame(X_pca,
                      columns=['PC1','PC2'])

pca_df["Cluster"] = clusters
```

PCA Scatterplot

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=pca_df,
    x='PC1',
    y='PC2',
    hue='Cluster',
    palette='Set1',
    s=80
)

plt.title("PCA Projection of Country Clusters")

plt.show()
```

Cluster Summary

```python
cluster_summary = df.groupby("KMeans_Cluster")[numeric_cols].mean()

cluster_summary
```
